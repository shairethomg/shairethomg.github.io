##Decision surface

Perform the classification of points using a decision surface using machine learning.


```python
# you need these librarys

import keras as kr
from keras.models import Sequential
from keras.layers import Dense, Input

import numpy as np #array manipulation and mathematical operations in Python.
import matplotlib.pyplot as plt #Creating 2D graphics and visualizations.
from sklearn.datasets import make_circles #generates a synthetic data set, typically used for classification problems.
import matplotlib as mpl
```

```python
n=500 #number of data
p=2 #categories

X,y = make_circles(n_samples=n,factor=0.4,noise=0.1)
Xval,yval = make_circles(n_samples=20,factor=0.4,noise=0.1)
y = y[:,np.newaxis]
cmap = mpl.colors.ListedColormap(['blue', 'yellow'])
scatter = plt.scatter(X[:,0],X[:,1],c=y.flatten())
plt.colorbar(scatter, ticks=[0, 1])
plt.plot(Xval[:,0],Xval[:,1],'k+', label="Validation Data")
plt.legend()
plt.show()
```

```python
#You can add or remove hidden layers, use any number of neurons, you are free to enjoy your freedom.

model = kr.Sequential()


model.add(Input(shape=(2,))) # model expects rows of data with 2 variables (x,y) nn[0]

model.add(kr.layers.Dense(15,activation='relu'))

model.add(kr.layers.Dense(1,activation='sigmoid')) #output

print(model.summary() )
```

```python
#Define network parameters

lr = 0.1
epochs = 25
loss="binary_crossentropy"
optimizer1=kr.optimizers.SGD(learning_rate=lr)
optimizer2=kr.optimizers.Adam(learning_rate=lr)
optimizer3=kr.optimizers.RMSprop(learning_rate=lr)

model.compile(loss="binary_crossentropy", optimizer=optimizer2, metrics=['acc'])
```

```python
#Training the network

history=model.fit(X,y.flatten(),epochs=epochs, validation_split = 0.2)
print(history.history.keys())
```

```python
#Graphing loss and accuracy function

plt.plot(history.history['loss'], 'b--', label='loss_training', linewidth=1)
plt.plot(history.history['val_loss'], 'b-', label='loss_val', linewidth=2)
plt.plot(history.history['acc'], 'r--', label='accuracy_training', linewidth=1)
plt.plot(history.history['val_acc'], 'r-', label='val_accuracy', linewidth=2)
plt.xlabel('Epochs')
plt.ylabel('Loss / Accuracy')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Evaluate the model

test_loss, test_acc = model.evaluate(Xval,yval,verbose=0)
print('\nTest accuracy:', test_acc)
```

Let's create a mesh of points in a two-dimensional space and use the trained model to make predictions at those points.

```python
#Making predictions

res=50
_x0 = np.linspace(X[:,0].min(), X[:,0].max(), res)
_x1 = np.linspace(X[:,1].min(), X[:,1].max(), res)

XX,XY = np.meshgrid(_x0,_x1)
XXgridval=np.stack((XX.flatten(),XY.flatten()),1)

YYgridpred = model.predict(XXgridval)
YYgridpred=YYgridpred.reshape((res,res))
```

```python
#Graph the results

plt.figure(figsize=(10,5))
plt.subplot(121)
plt.pcolormesh(_x0, _x1, YYgridpred, cmap="jet")
plt.scatter(X[:,0], X[:,1], c=y[:,0],cmap="jet",edgecolors='k')
plt.scatter(Xval[:,0], Xval[:,1], c=yval,cmap="jet",edgecolors='gray')
plt.xlim(X[:,0].min(), X[:,0].max())
plt.ylim(X[:,1].min(), X[:,1].max())
plt.colorbar()
ax1 = plt.subplot(122)
ax1.plot(history.history['val_acc'],'b',label='acc')
ax1.set_ylabel('Accuracy', color='b')
ax1.tick_params(axis='y', labelcolor='b')
ax2 = ax1.twinx()
ax2.plot(history.history['val_loss'],'r',label='loss')
ax2.set_ylabel('Loss function', color='r')
ax2.tick_params(axis='y', labelcolor='r')
plt.title(f'epoch:{epochs} Test accuracy {test_acc:0.3f}')
```
